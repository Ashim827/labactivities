```python
import random
import time

print("⚡ REACTION TIME TESTER ⚡")
print("Press ENTER as fast as you can when you see 'GO!'...")
time.sleep(2)

# Random wait time between 2 to 5 seconds
print("\nGet ready...")
time.sleep(random.randint(2, 5))

start_time = time.time()
input("🔴 GO! GO! GO! (Press ENTER now!)")
end_time = time.time()

# Calculate reaction speed
reaction_time = round(end_time - start_time, 3)
print(f"\n⚡ Your reaction time: {reaction_time} seconds!")
```
